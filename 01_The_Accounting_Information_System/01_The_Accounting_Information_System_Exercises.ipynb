##### **E2.2 (LO 2) (Corrected Trial Balance) The following trial balance of Wanda Landowska Company does not balance. Your review of the ledger reveals the following.**

- (a) Each account had a normal balance. 
- (b) The debit footings in Prepaid Insurance, Accounts Payable, and Property Tax Expense were each understated 100. 
- (c) A transposition error was made in Accounts Receivable and Service Revenue; the correct balances for Accounts Receivable and Service Revenue are 2,750 and 6,690, respectively.
- (d) A debit posting to Advertising Expense of 300 was omitted. 
- (e) A 1,500 cash drawing by the owner was debited to Owner’s Capital and credited to Cash.


| Wanda Landowska Company |
| :---: |
| Trial Balance |
| April 30, 2025 |
| ----------------------------------------------------- |

| | Debit | Credit |
| --- | :---: | :---: |
| Cash | 4,800 | |
| Accounts Receivable | 2,570 | |
| Prepaid Insurance | 700 | |

| Accounts Payable |  | 4,500 |

|  | **20,890** | **24,500** |

##### **Instruction**

Prepare a correct trial balance.

```python
tb = TrialBalance()
tb.reset()
tb.add_account("Cash", debit=4800, account_type="asset")
tb.add_account("Accounts Receivable", debit=2570, account_type="asset")
tb.add_account("Prepaid Insurance", debit=700, account_type="asset")
tb.add_account("Equipment", debit=8000, account_type="asset")
tb.add_account("Accounts Payable", credit=4500, account_type="liability")
tb.add_account("Property Taxes Payable", credit=560, account_type="liability")
tb.add_account("Owner's Capital", credit=11200, account_type="equity")
tb.add_account("Service Revenue", credit=6960, account_type="revenue")
tb.add_account("Salaries and Wages Expense", debit=4200, account_type="expense")
tb.add_account("Advertising Expense", debit=1100, account_type="expense")
tb.add_account("Property Tax Expense", debit=800, account_type="expense")
```

```python
display(HTML(format_trial_balance_table(tb).to_html(index=False)))
print("Balanced:", tb.is_balanced())
```

##### **Adjustments**

```python
tb.adjust_account("Prepaid Insurance", debit_change=100)      # (b)
tb.adjust_account("Accounts Payable", credit_change=-100)     # (b)
tb.adjust_account("Property Tax Expense", debit_change=100)   # (b)
tb.adjust_account("Accounts Receivable", debit_change=180)    # (c)
tb.adjust_account("Service Revenue", credit_change=-270)      # (c)
tb.adjust_account("Advertising Expense", debit_change=300)    # (d)


je_corrections = Ledger()
je_corrections.add_entry(
    [
        {"date": "2025-04-30", "account": "Owner's Drawings", "debit": 1500, "credit": 0},
        {"date": "2025-04-30", "account": "Owner's Capital", "debit": 0, "credit": 1500}
    ],
    description="Reclassify $1,500 cash drawing from Owner's Capital to Owner's Drawings"
)

tb.apply_ledger(je_corrections, account_types={"Owner's Drawings": "equity"})
```

```python
display(HTML(format_trial_balance_table(tb).to_html(index=False)))
print("Balanced:", tb.is_balanced())
```

##### **E2.3 (LO 2) (Corrected Trial Balance) The following trial balance of Blues Traveler Corporation does not balance.**

| Blues Traveler Corporation |
| :---: |
| Trial Balance |
| April 30, 2025 |
| -------------------------------------------- |

| | Debit | Credit |
| --- | :---: | :---: |
| Cash | 5,912 | |
| Accounts Receivable | 5,240 | |
| Supplies | 2,967 | |
| Equipment | 6,100 | |
| Accounts Payable |  | 7,044 |
| Common Stock |  | 8,000 |
| Retained Earnings |  | 2,000 |
| Service Revenue |  | 5,200  |
| Office Expense | 4,320 |  |
|  | **24,539** | **22,244** |

An examination of the ledger shows these errors.

1. Cash received from a customer on account was recorded (both debit and credit) as 1,380 instead of 1,830.
2. The purchase on account of a computer costing 3,200 was recorded as a debit to Office Expense and a credit to Accounts Payable.
3. Services were performed on account for a client, 2,250, for which Accounts Receivable was debited 2,250 and Service Revenue was credited 225.
4. A payment of 95 for telephone charges was entered as a debit to Office Expense and a debit to Cash.
5. The Service Revenue account was totaled at 5,200 instead of 5,280.

##### **Instruction**

From this information prepare a corrected trial balance.

```python
blue_tb = TrialBalance()
blue_tb.reset()
blue_tb.add_account("Cash", debit=5912, account_type="asset")
blue_tb.add_account("Accounts Receivable", debit=5240, account_type="asset")
blue_tb.add_account("Supplies", debit=2967, account_type="asset")
blue_tb.add_account("Equipment", debit=6100, account_type="asset")
blue_tb.add_account("Accounts Payable", credit=7044, account_type="liability")
blue_tb.add_account("Common Stock", credit=8000, account_type="equity")
blue_tb.add_account("Retained Earnings", credit=2000, account_type="equity")
blue_tb.add_account("Service Revenue", credit=5200, account_type="revenue")
blue_tb.add_account("Office Expense", debit=4320, account_type="expense")
```

```python
display(HTML(format_trial_balance_table(blue_tb).to_html(index=False)))
print("Balanced:", blue_tb.is_balanced())
```

##### **Adjustments**

```python
blue_ledger = Ledger()
blue_ledger.reset()
blue_ledger.add_entry(
    [
        {"date": "2025-04-30", "account": "Cash", "debit": 1830-1380, "credit": 0},
        {"date": "2025-04-30", "account": "Accounts Receivable", "debit": 0, "credit": 1830-1380}
    ],
    description="Error 1: Correction of erroneous journal entry"
)

blue_ledger.add_entry(
    [
        {"date": "2025-04-30", "account": "Equipment", "debit": 3200, "credit": 0},
        {"date": "2025-04-30", "account": "Office Expense", "debit": 0, "credit": 3200}
    ],
    description="Error 2: Reclassify equipment recorded as expense"
)
blue_tb.apply_ledger(blue_ledger)
```

```python
display(HTML(format_trial_balance_table(blue_tb).to_html(index=False)))
print("Balanced:", blue_tb.is_balanced())
```
