```python
import os
import sys
import pandas as pd
from datetime import date
from IPython.display import display, HTML

mod_path = os.path.abspath(os.path.join(".."))
if mod_path not in sys.path:
    sys.path.append(mod_path)

from ledger import Ledger

def months_elapsed(payment_date: date, entry_date: date) -> int:
    """Used to get the number of months elapsed"""
    return (entry_date.year - payment_date.year) * 12 + (entry_date.month - payment_date.month) + 1 

def format_ledger_table(ledger, currency=True):
    """
    Takes a Ledger, returns a formatted DataFrame with a TOTAL row,
    ready to display with display(HTML(...)).
    Set currency=False to keep raw numbers (e.g., for further calculations).
    """
    table = ledger.to_table()

    totals = pd.DataFrame([{
        "txn_id": "",
        "date": "",
        "account": "TOTAL",
        "debit": table["debit"].sum(),
        "credit": table["credit"].sum(),
        "description": "",
    }])
    table_with_total = pd.concat([table, totals], ignore_index=True)

    if currency:
        table_with_total["debit"] = table_with_total["debit"].apply(
            lambda x: f"{x:,.2f}" if x != "" and x != 0 else ""
        )
        table_with_total["credit"] = table_with_total["credit"].apply(
            lambda x: f"{x:,.2f}" if x != "" and x != 0 else ""
        )

    table_with_total.columns = [col.replace("_", " ").title() for col in table_with_total.columns]

    return table_with_total
```

##### **BE2.12 (LO 6) Kelly Company had cash receipts from customers in 2025 of 142,000. Cash payments for operating expenses were 97,000. Kelly has determined that at January 1, accounts receivable was 13,000, and prepaid expenses were 17,500. At December 31, accounts receivable was 18,600, and pre-paid expenses were 23,200. Compute (a) service revenue and (b) operating expenses.**

```python
# It is easier if we just define these functions right away.

def gaap_revenue(cash_receipts, 
                 beginning_ar, 
                 ending_ar,
                 beginning_unearned_revenues,
                 ending_unearned_revenues):

    return cash_receipts - beginning_ar + ending_ar + beginning_unearned_revenues - ending_unearned_revenues


def gaap_expenses(cash_payments,
                  beginning_prepaids,
                  ending_prepaids,
                  beginning_accrued_expenses,
                  ending_accrued_expenses):

    return cash_payments + beginning_prepaids - ending_prepaids + beginning_accrued_expenses - ending_accrued_expenses
```

```python
cash_receipts = 142000
cash_payments = 97000
beginning_ar = 13000
ending_ar = 18600
beginning_unearned_revenues = 0
ending_unearned_revenues = 0
beginning_prepaids = 17500
ending_prepaids = 23200
beginning_accrued_expenses = 0
ending_accrued_expenses = 0

service_revenue = gaap_revenue(cash_receipts, 
                               beginning_ar, ending_ar, 
                               beginning_unearned_revenues, 
                               ending_unearned_revenues)

operating_expenses = gaap_expenses(cash_payments,
                                   beginning_prepaids,
                                   ending_prepaids,
                                   beginning_accrued_expenses,
                                   ending_accrued_expenses)

net_income = service_revenue - operating_expenses

print(f"{"Service Revenue":<20} | {service_revenue:>10,.0f}")
print(f"{"Operating Expenses":<20} | {operating_expenses:>10,.0f}")
print(f"{"-"*20:<20} | {"-"*10:>10}")
print(f"{"Net Income":<20} | {net_income:>10,.0f}")
```
