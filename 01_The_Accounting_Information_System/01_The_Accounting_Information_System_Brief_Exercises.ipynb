```python
import os
import sys
import pandas as pd
from datetime import date
from IPython.display import display, HTML

mod_path = os.path.abspath(os.path.join(".."))
if mod_path not in sys.path:
    sys.path.append(mod_path)

from ledger import Ledger

def months_elapsed(payment_date: date, entry_date: date) -> int:
    """Used to get the number of months elapsed"""
    return (entry_date.year - payment_date.year) * 12 + (entry_date.month - payment_date.month) + 1 
```

##### **BE2.2 (LO 2) Agazzi Repair Shop had the following transactions during the first month of business as a proprietorship. Journalize the transactions. (Omit explanations.)**

- **Aug 2 - Invested 12,000 cash and 2,500 of equipment in the business.**
- **Aug 7 - Purchased supplies on account for 500. (Debit asset account.)**
- **Aug 12 - Performed services for clients, for which 1,300 was collected in cash and 670 was billed to the clients.**
- **Aug 15 - Paid August rent 600.**
- **Aug 19 - Counted supplies and determined that only 270 of the supplies purchased on August 7 are still on hand.**

```python
agazzi = Ledger()
agazzi.reset()

cash = 12000
equipment = 2500
investment = cash + equipment

aug_2 = agazzi.add_entry(
    [
        {"date": "2025-8-2", "account": "Cash", "debit": cash, "credit": 0},
        {"date": "2025-8-2", "account": "Equipment", "debit": equipment, "credit": 0},
        {"date": "2025-8-2", "account": "Owner's Capital", "debit": 0, "credit": investment},
    ],
    description="",
)

supplies_purchased = 500

aug_7 = agazzi.add_entry(
    [
        {"date": "2025-8-7", "account": "Supplies", "debit": supplies_purchased, "credit": 0},
        {"date": "2025-8-7", "account": "Accounts Payable", "debit": 0, "credit": supplies_purchased},
    ],
    description="",
)

cash = 1300
ar = 670
revenue = cash + ar

aug_12 = agazzi.add_entry(
    [
        {"date": "2025-8-12", "account": "Cash", "debit": cash, "credit": 0},
        {"date": "2025-8-12", "account": "Accounts Receivable", "debit": ar, "credit": 0},
        {"date": "2025-8-12", "account": "Service Revenue", "debit": 0, "credit": revenue},
    ],
    description="",
)
aug_15 = agazzi.add_entry(
    [
        {"date": "2025-8-15", "account": "Rent Expense", "debit": 600, "credit": 0},
        {"date": "2025-8-15", "account": "Cash", "debit": 0, "credit": 600},
    ],
    description="",
)

supplies_on_hand = 270
supplies_expense = supplies_purchased - supplies_on_hand

aug_19 = agazzi.add_entry(
    [
        {"date": "2025-8-19", "account": "Supplies Expense", "debit": supplies_expense, "credit": 0},
        {"date": "2025-8-19", "account": "Supplies", "debit": 0, "credit": supplies_expense},
    ],
    description="",
)
```

```python

agazzi.print_transaction()
```

##### **BE2.3 (LO 2, 3) On July 1, 2025, Crowe Co. pays 15,000 to Zubin Insurance Co. for a 3-year insurance policy. Both companies have fiscal years ending December 31. For Crowe Co., journalize the entry on July 1 and the adjusting entry on December 31.**

```python
crowe = Ledger()
crowe.reset()

insurance = 15000
expense_per_year = insurance / 3

jul_1 = crowe.add_entry(
    [
        {"date": "2025-7-2", "account": "Pre-Paid Insurance", "debit": insurance, "credit": 0},
        {"date": "2025-7-2", "account": "Cash", "debit": 0, "credit": insurance},
    ],
    description="",
)

payment_date = date(2025, 7, 1)
adjusting_date = date(2025, 12, 31)
months = months_elapsed(payment_date, adjusting_date)
insurance_expense = (expense_per_year / 12) * months

dec_31 = crowe.add_entry(
    [
        {"date": "2025-7-2", "account": "Insurance Expense", "debit": insurance_expense, "credit": 0},
        {"date": "2025-7-2", "account": "Pre-Paid Insurance", "debit": 0, "credit": insurance_expense},
    ],
    description="",
)
```

```python
entries = crowe.to_table()
entries["debit"] = entries["debit"].apply(lambda x: f"{x:,.2f}")
entries["credit"] = entries["credit"].apply(lambda x: f"{x:,.2f}")
display(HTML(entries.to_html(index=False)))
```
