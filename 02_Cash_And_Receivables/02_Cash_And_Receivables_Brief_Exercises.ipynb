##### **BE6.2 (LO 2) Restin Co. uses the gross method to record sales made on credit. On June 1, 2025, it made sales of 50,000 with terms 3/15, n/45. On June 12, 2025, Restin received full payment for the June 1 sale. Prepare the required journal entries for Restin Co**

```python
discount = .03
sales = 50000
discounted_sales = 0
sales_discount = 0
sales_date = date(2025, 6, 1)
payment_date = date(2025, 6, 12)
days = (payment_date - sales_date).days

if days <= 15:
    sales_discount = sales * discount 
    discounted_sales = sales - sales_discount

restin = Ledger()
restin.reset()

restin.add_entry(
    [
        {"date": "2025-6-1", "account": "Accounts Receivable", "debit": sales, "credit": 0},
        {"date": "2025-6-1", "account": "Sales Revenue", "debit": 0, "credit": sales},
    ],
    description="",
)

restin.add_entry(
    [
        {"date": "2025-6-12", "account": "Cash", "debit": discounted_sales, "credit": 0},
        {"date": "2025-6-12", "account": "Sales Discounts", "debit": sales_discount, "credit": 0},
        {"date": "2025-6-12", "account": "Accounts Receivable", "debit": 0, "credit": sales},
    ],
    description="",
)

display(HTML(format_ledger_table(restin).to_html(index=False)))
```

##### **BE6.4 (LO 2) Roeher Company sold 9,000 of its specialty shelving to Elkins Office Supply Co. on account.**

**Prepare the entries when:**

##### **BE6.7 (LO 4) Milner Family Importers sold goods to Tung Decorators for 30,000 on November 1, 2025, accepting Tung’s 30,000, 6-month, 6% note. Prepare Milner’s November 1 entry, December 31 annual adjusting entry, and May 1 entry for the collection of the note and interest.**

```python
note = 30000
stated_rate = .06
periods = 6
interest = note * (stated_rate * (6/12))

milner = Ledger()
milner.reset()

milner.add_entry(
    [
        {"date": "2025-11-1", "account": "Notes Receivable", "debit": note, "credit": 0},
        {"date": "2025-11-1", "account": "Sales Revenue", "debit": 0, "credit": note},
    ],
    description="",
)

interest_for_the_year = interest * (2 / 6)

milner.add_entry(
    [
        {"date": "2025-12-31", "account": "Interest Receivable", "debit": interest_for_the_year, "credit": 0},
        {"date": "2025-12-31", "account": "Interest Revenue", "debit": 0, "credit": interest_for_the_year},
    ],
    description="",
)
cash = note + interest
interest_revenue = interest * (4 / 6)
milner.add_entry(
    [
        {"date": "2026-5-1", "account": "Cash", "debit": cash, "credit": 0},
        {"date": "2026-5-1", "account": "Notes Receivable", "debit": 0, "credit": note},
        {"date": "2026-5-1", "account": "Interest Receivable", "debit": 0, "credit": interest_for_the_year},
        {"date": "2026-5-1", "account": "Interest Revenue", "debit": 0, "credit": interest_revenue},
    ],
    description="",
)

display(HTML(format_ledger_table(milner).to_html(index=False)))
```

##### **BE6.10 (LO 5) Wood Incorporated factored 150,000 of accounts receivable with Engram Factors Inc. on a without-recourse basis. Engram assesses a 2% finance charge of the amount of accounts receivable and retains an amount equal to 6% of accounts receivable for possible adjustments. Prepare the journal entry for Wood Incorporated and Engram Factors to record the factoring of the accounts receivable to Engram.**

```python
finance_charge_pct = .02
retainer_pct = .06
accounts_receivable = 150000

retainer = accounts_receivable * retainer_pct
finance_charge = accounts_receivable * finance_charge_pct

cash = f'{accounts_receivable} {retainer} - {finance_charge} -'
cash = RPNEngine(cash).safe_evaluate().value
```

**Wood Incorporated**

```python
wood = Ledger()
wood.reset()

wood.add_entry(
    [
        {"date": "2025-1-1", "account": "Cash", "debit": cash, "credit": 0},
        {"date": "2025-1-1", "account": "Receivable from Factor (Engram)", "debit": retainer, "credit": 0},
        {"date": "2025-1-1", "account": "Loss on Sale of Receivables", "debit": finance_charge, "credit": 0},
        {"date": "2025-1-1", "account": "Accounts Receivables", "debit": 0, "credit": accounts_receivable},
    ],
    description="",
)

display(HTML(format_ledger_table(wood).to_html(index=False)))
```

**Engram Factors**

```python
engram = Ledger()
engram.reset()

engram.add_entry(
    [
        {"date": "2025-1-1", "account": "Accounts Receivable", "debit": accounts_receivable, "credit": 0},
        {"date": "2025-1-1", "account": "Due to Customer (Wood)", "debit": 0, "credit": retainer},
        {"date": "2025-1-1", "account": "Interest Revenue", "debit": 0, "credit": finance_charge},
        {"date": "2025-1-1", "account": "Cash", "debit": 0, "credit": cash},
    ],
    description="",
)

display(HTML(format_ledger_table(engram).to_html(index=False)))
```

##### **BE6.12 (LO 5) Arness Woodcrafters sells 250,000 of receivables to Commercial Factors, Inc. on a with recourse basis. Commercial assesses a finance charge of 5% and retains an amount equal to 4% of accounts receivable. Arness estimates the fair value of the recourse liability to be 8,000. Prepare the journal entry for Arness to record the sale.**

```python
arness = Ledger()
arness.reset()

finance_charge_pct = .05
retainer_pct = .04
accounts_receivable = 250000

retainer = accounts_receivable * retainer_pct
finance_charge = accounts_receivable * finance_charge_pct

cash = f'{accounts_receivable} {retainer} - {finance_charge} -'
cash = RPNEngine(cash).safe_evaluate().value

recourse_liability = 8000
loss_on_sale = f'{accounts_receivable} {cash} {retainer} + {recourse_liability} - -'
loss_on_sale = RPNEngine(loss_on_sale).safe_evaluate().value

arness.add_entry(
    [
        {"date": "2025-1-1", "account": "Cash", "debit": cash, "credit": 0},
        {"date": "2025-1-1", "account": "Receivable from Seneca", "debit": retainer, "credit": 0},
        {"date": "2025-1-1", "account": "Loss on Sale of Receivables", "debit": loss_on_sale, "credit": 0},
        {"date": "2025-1-1", "account": "Accounts Receivables", "debit": 0, "credit": accounts_receivable},
        {"date": "2025-1-1", "account": "Recourse Liability", "debit": 0, "credit": recourse_liability},
    ],
    description="",
)

display(HTML(format_ledger_table(arness).to_html(index=False)))
```
