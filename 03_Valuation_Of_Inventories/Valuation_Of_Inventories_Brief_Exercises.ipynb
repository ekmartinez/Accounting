##### **BE7.2 (LO 1) Matlock Company uses a perpetual inventory system. Its beginning inventory consists of 50 units that cost 34 each. During June, the company purchased 150 units at 34 each, returned 6 units for credit, and sold 125 units at 50 each. Journalize the June transactions.**

```python
matlock = Ledger()
matlock.reset()

cost = 34
price = 50

purchases = 150 * cost
credits = 6 * price
sales = 125 * price
cogs = 125 * cost

matlock.add_entry(
    [
        {"date": "2025-6-30", "account": "Inventory", "debit": purchases, "credit": 0},
        {"date": "2025-6-30", "account": "Accounts Payable", "debit": 0, "credit": purchases},
    ],
    description="",
)

matlock.add_entry(
    [
        {"date": "2025-6-30", "account": "Accounts Payable", "debit": credits, "credit": 0},
        {"date": "2025-6-30", "account": "Inventory", "debit": 0, "credit": credits},
    ],
    description="",
)

matlock.add_entry(
    [
        {"date": "2025-6-30", "account": "Accounts Receivable", "debit": sales, "credit": 0},
        {"date": "2025-6-30", "account": "Sales Revenue", "debit": 0, "credit": sales},
        {"date": "2025-6-30", "account": "Cost of Goods Sold", "debit": cogs, "credit": 0},
        {"date": "2025-6-30", "account": "Inventory", "debit": 0, "credit": cogs},
    ],
    description="",
)

display(HTML(format_ledger_table(matlock).to_html(index=False)))
```

##### **BE7.3 (LO 2) Stallman Company took a physical inventory on December 31 and determined that goods costing 200,000 were on hand. Not included in the physical count were 25,000 of goods purchased from Pelzer Corporation, f.o.b. shipping point, and 22,000 of goods sold to Alvarez Company for 30,000, f.o.b. destination. Both the Pelzer purchase and the Alvarez sale were in transit at year-end. What amount should Stallman report as its December 31 inventory?**

```python
physical_inventory = 200000
purchases_in_transit = 25000 # FOB Shipping Point
sales_in_transit = 22000     # FOB Destination
correct_inventory = physical_inventory + purchases_in_transit + sales_in_transit
print(f"Stallman should report $ {correct_inventory:,.0f} as its December 31 inventory.\n")

print(f"{"Physical Inventory":<26} {physical_inventory:>10,.0f}")
print(f"{"Add: Purchases in Transit":<26} {purchases_in_transit:>10,.0f}")
print(f"{"Add: Sales in Transit":<26} {sales_in_transit:>10,.0f}")
print(f"{"Correct Inventory":<26} {correct_inventory:>10,.0f}")
```

##### **BE7.4 (LO 3) Amsterdam Company uses a periodic inventory system. For April, when the company sold 600 units, the following information is available.**

| | Units | Unit Cost | Total Cost |
| --- | ---: | :---: | ---: |
| April 1 inventory | 250 | 10 | 2,500 |
| April 15 purchase | 400 | 12 | 4,800 |
| April 23 purchase | 350 | 13 | 4,550 |
| **Totals** | **1,000** |  | **11,850** |

Compute the April 30 inventory and the April cost of goods sold using the average-cost method.

```python
sales_units = 600
afs_dollars = 11850
afs_units = 1000
weighted_average_cost = afs_dollars / afs_units
sales_dollars = sales_units * weighted_average_cost
ending_inventory = afs_dollars - sales_dollars
cogs = afs_dollars - ending_inventory

print("Average-Cost Method")
print("-------------------")
print(f"{"Inventory":<26} $ {ending_inventory:>10,.0f}")
print(f"{"Cost of Goods Sold":<26} $ {cogs:>10,.0f}")
```
